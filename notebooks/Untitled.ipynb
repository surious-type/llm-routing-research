# Вероятность выбора семантического маршрута LLM

## Аннотация

Как из `logprobs` получить **относительную вероятность выбора намерения** и почему одной такой вероятности недостаточно для надёжной маршрутизации.

Исследуется бинарная задача:

- `CONTINUE` — новое сообщение продолжает текущую пользовательскую задачу;
- `NEW` — новое сообщение вводит новую независимую задачу.

## Постановка

По истории диалога и новому сообщению определить, это `CONTINUE` текущей задачи или `NEW` — новая самостоятельная задача.
